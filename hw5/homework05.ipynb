## Answer 1

### 1. Probability that an individual goes snorkling and has lunch at the restaurant

There are 200 individuals who go snorkling and have lunch at the restaurant, out of 600 total individuals.  
So, the probability is:  
$P = \frac{200}{600} = \frac{1}{3} \approx 0.33$

### 2. Probability that an individual does *not* have lunch at the restaurant

There are 450 individuals who have lunch at the restaurant, so the number who do *not* is $600 - 450 = 150$.  
So, the probability is:  
$P = \frac{150}{600} = \frac{1}{4} = 0.25$

### 3. Probability that an individual goes snorkling **or** goes dancing

Use inclusion-exclusion:
- Number who go snorkling = 300
- Number who go dancing = 200
- Number who do both = 100

So, the total is $300 + 200 - 100 = 400$.

Probability is:  
$P = \frac{400}{600} = \frac{2}{3} \approx 0.67$

### 4. Probability that an individual has lunch but does *not* go snorkling

Number who have lunch = 450  
Number who have lunch **and** go snorkling = 200  
So, those who have lunch but *do not* go snorkling = $450 - 200 = 250$.

Probability is:  
$P = \frac{250}{600} \approx 0.417$

### 5. Probability that an individual goes dancing **given** they have lunch

We are asked: $P(\text{dancing} \mid \text{lunch})$.

Number who have lunch = 450  
Number who have lunch and go dancing = 100

So, the conditional probability is:  
$P = \frac{100}{450} \approx 0.222$
