What is the probability $P(D \mid T)$ that they have the disease given that they get tested?


2. Suppose the test is not always accurate. In particular, the probability $P(pos \mid D)$ of a positive test result given someone has the disease is 95% or 0.95. The probability $P(pos \mid \overline{D})$ of a positive test result given someone does **not** have the disease is 10% or 0.1. Suppose a random person takes the test. What is the probability $P(pos)$ that they test positive?


3. What is the probability $P(D \mid pos)$ that a random person has the disease given that they test positive?
